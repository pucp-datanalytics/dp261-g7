# Sprint 1: Business Case & Analytics KPIs
**Proyecto: Retención de Talento (HR Attrition)**

Este notebook define el problema de negocio y establece las métricas de éxito del modelo predictivo.

### KPIs de Negocio
1. **Reducción de Rotación:** Disminuir la tasa de fuga del 16% al 12% en 12 meses.
2. **Ahorro Operativo:** Lograr un 20% de ahorro anual en costos de reclutamiento.

### Métricas de Machine Learning
- **Métrica Principal:** **Recall (Sensibilidad)** de la clase 1 (Fuga) > 80%, debido al alto costo de no detectar a empleados en riesgo.
- **Métricas Secundarias:** F1-Score y ROC-AUC (> 0.75) para evaluar la capacidad de discriminación del modelo.
