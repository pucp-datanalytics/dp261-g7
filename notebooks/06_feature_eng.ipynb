# Sprint 2: Feature Engineering

Creación de variables sintéticas derivadas.

```python
import pandas as pd
from pathlib import Path

df = pd.read_csv("../data/interim/employee_performance_clean.csv")

# 1. Ratio Salario / Edad
df["ratio_salario_edad"] = df["monthly_salary"] / df["age"]

# 2. Antigüedad x Satisfacción
df["antiguedad_satisfaccion"] = df["years_at_company"] * df["job_satisfaction"]

# 3. Rango Edad
df["rango_edad"] = pd.cut(
    df["age"],
    bins=[0, 30, 45, 60, 100],
    labels=["joven", "adulto", "maduro", "senior"]
).astype(str)

Path("../data/processed").mkdir(parents=True, exist_ok=True)
df.to_csv("../data/processed/features_data.csv", index=False)
print("Feature Engineering completado y datos guardados.")
```
