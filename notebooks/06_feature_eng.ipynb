```python
import pandas as pd
import numpy as np
from pathlib import Path

# 1. Cargar datos limpios
df = pd.read_csv("../data/interim/employee_performance_clean.csv")
print(f"Shape inicial: {df.shape}")

# 2. Feature 1: ratio salario / edad
df["ratio_salario_edad"] = df["monthly_salary"] / df["age"]

# 3. Feature 2: antigüedad × satisfacción
df["antiguedad_satisfaccion"] = df["years_at_company"] * df["job_satisfaction"]

# 4. Feature 3: rango etario
df["rango_edad"] = pd.cut(
    df["age"],
    bins=[0, 30, 45, 60, 100],
    labels=["joven", "adulto", "maduro", "senior"]
)

# 5. Verificar nulos
print("\nNulos en features creadas:")
print(df[["ratio_salario_edad", "antiguedad_satisfaccion", "rango_edad"]].isnull().sum())

# 6. Guardar dataset con features
Path("../data/processed").mkdir(parents=True, exist_ok=True)
df.to_csv("../data/processed/features_data.csv", index=False)

print(f"\n✅ Dataset guardado en: data/processed/features_data.csv")
print(f"Shape final: {df.shape}")
```
